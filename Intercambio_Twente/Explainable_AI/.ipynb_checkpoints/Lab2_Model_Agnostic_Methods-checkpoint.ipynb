## Setup

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import shap

from sklearn.base import clone
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.inspection import PartialDependenceDisplay, permutation_importance
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, roc_auc_score
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.tree import DecisionTreeClassifier

from xai_lab_utils import group_seed, personal_data_path, your_patient_index
```

```python
# Fill in your group number. Everything else depends on it.
GROUP_NUMBER = 1  # TODO: your group number, e.g. 7

S = group_seed(GROUP_NUMBER)   # use S everywhere a random seed is needed
print("Group", GROUP_NUMBER, "- seed S =", S)
```

```python
# TODO: paste your code from Lab 1 here:
#   - reading the data file (Q1)
path = personal_data_path(GROUP_NUMBER)
print(path)

with open(path, encoding="utf-8") as f:
    for _ in range(3):
        print(f.readline().rstrip())

df = pd.read_csv(path)  # Standard sep, header and decimal will work fine here
df.info()
df.head()

#   - the train/test split and the preprocessor (Q3)
X = df.drop(columns="target").astype(float)   # float: needed for PDP in recent scikit-learn
y = df["target"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, stratify=y, random_state=S)

num_cols = ["age", "trestbps", "chol", "thalach", "oldpeak"]   # numerical attributes: will be scaled
cat_cols = ["cp", "restecg", "slope", "ca", "thal"]   # categorical attributes: will be one-hot encoded
bin_cols = ["sex", "fbs", "exang"]   # binary attributes: kept as they are

preprocessor = ColumnTransformer(
    [("num", StandardScaler(), num_cols),
     ("cat", OneHotEncoder(drop="first", sparse_output=False, handle_unknown="ignore"), cat_cols),
     ("bin", "passthrough", bin_cols)],
    verbose_feature_names_out=False,
).set_output(transform="pandas")

#   - the three fitted models: logreg, tree, knn
logreg = make_pipeline(clone(preprocessor), LogisticRegression(max_iter=1000))
tree = DecisionTreeClassifier(random_state=S)            # a tree needs no scaling
knn = make_pipeline(clone(preprocessor), KNeighborsClassifier(n_neighbors=5))

models = {"logistic": logreg, "tree": tree, "kNN": knn}

for name, model in models.items():
    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    accuracy = accuracy_score(y_test, y_pred)
    roc_auc = roc_auc_score(y_test, y_prob)

    print(f"{name}:")
    print(f"  Accuracy: {accuracy:.4f}")
    print(f"  ROC-AUC:  {roc_auc:.4f}")
```

```python
rf = RandomForestClassifier(n_estimators=300, random_state=S).fit(X_train, y_train)
models = {"logistic": logreg, "tree": tree, "kNN": knn, "random forest": rf}

p = your_patient_index(GROUP_NUMBER, len(X_test))
x_p = X_test.iloc[[p]]
```

## Q1) Permutation importance

a) **[Basic]** Before you run it, write down which three attributes you expect to have the highest permutation importance for the logistic model (look at your odds ratios from Lab 1). Then perform permutation importance for the logistic model for all attributes and plot the results. Give a brief interpretation. Was your expectation right?

b) **[Basic]** Which model is most likely to show a bigger difference in prediction when using permutation importance? Run it for all models and make a table with the top 5 attributes of each model.

c) **[Regular]** Run permutation importance for the random forest on the training set and on the test set. What is the difference? What does it tell you about the model?

d) **[Advanced]** Add a copy of `thalach` with a little noise to the data and run permutation importance again. What happens to the importance of `thalach`? Why?

e) **[Advanced]** Add a column of random numbers. Compare the built-in `feature_importances_` of the random forest with permutation importance. Which one gives importance to the random column, and why?

> **Hint:** `permutation_importance(model, X_test, y_test, n_repeats=30, random_state=S)`; `r.importances_mean`, `r.importances_std`

```python
r = permutation_importance(logreg, X_test, y_test, n_repeats=30, random_state=S)

imp = pd.DataFrame({"mean": r.importances_mean, "std": r.importances_std},
                   index=X_test.columns).sort_values("mean")
plt.figure(figsize=(7, 6))
plt.barh(imp.index, imp["mean"], xerr=imp["std"])
plt.xlabel("decrease in accuracy")
plt.title(f"Permutation importance, logistic model - group {GROUP_NUMBER}")
plt.show()

# TODO (b): run it for all models and make a table with the top 5 of each model
models_imps = {}

for name, model in models.items():
    r = permutation_importance(
        model,
        X_test,
        y_test,
        n_repeats=30,
        random_state=S
    )

    imp = pd.DataFrame({
        "mean": r.importances_mean,
        "std": r.importances_std
    }, index=X_test.columns)

    models_imps[name] = imp.sort_values("mean", ascending=False).head(5)

# Create a table
top5 = pd.concat(models_imps, names=["model", "attribute"])

print("\n================================ Top-5 highest permutation importances for each model ================================")
print(top5)

# TODO (c): random forest on the training set and on the test set
rf = RandomForestClassifier(
    n_estimators=300,
    random_state=S
).fit(X_train, y_train)

r_train = permutation_importance(
    rf, X_train, y_train,
    n_repeats=30,
    random_state=S
)

imp_train = pd.DataFrame({
    "mean": r_train.importances_mean,
    "std": r_train.importances_std
}, index=X_train.columns).sort_values("mean", ascending=False)

r_test = permutation_importance(
    rf, X_test, y_test,
    n_repeats=30,
    random_state=S
)

imp_test = pd.DataFrame({
    "mean": r_test.importances_mean,
    "std": r_test.importances_std
}, index=X_test.columns).sort_values("mean", ascending=False)

print("\n================================ Permutation importance for random forest with training and test sets ================================")
print("Training set")
print(imp_train)

print("\nTest set")
print(imp_test)

# TODO (d): add a copy of talach with noise and repeat
X_noise = X.copy()

# Add a noisy copy of thalach
rng = np.random.RandomState(S)
X_noise["thalach_noisy"] = X_noise["thalach"] + rng.normal(
    0, 1, size=len(X_noise)
)

X_train_n, X_test_n, y_train_n, y_test_n = train_test_split(
    X_noise, y, test_size=0.25, stratify=y, random_state=S
)

rf_noise = RandomForestClassifier(
    n_estimators=300,
    random_state=S
).fit(X_train_n, y_train_n)

r_noise = permutation_importance(
    rf_noise,
    X_test_n,
    y_test_n,
    n_repeats=30,
    random_state=S
)

imp_noise = pd.DataFrame({
    "mean": r_noise.importances_mean,
    "std": r_noise.importances_std
}, index=X_test_n.columns).sort_values("mean", ascending=False)

print("\n================================ Permutation importance with copy of talach with noise ================================")
print(imp_noise)

# TODO (e): add a column of random values and repeat
X_random = X.copy()

rng = np.random.RandomState(S)
X_random["random"] = rng.rand(len(X_random))

X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(
    X_random, y, test_size=0.25, stratify=y, random_state=S
)

rf_random = RandomForestClassifier(
    n_estimators=300,
    random_state=S
).fit(X_train_r, y_train_r)

built_in = pd.Series(
    rf_random.feature_importances_,
    index=X_random.columns
).sort_values(ascending=False)

r_random = permutation_importance(
    rf_random,
    X_test_r,
    y_test_r,
    n_repeats=30,
    random_state=S
)

permutation = pd.Series(
    r_random.importances_mean,
    index=X_random.columns
).sort_values(ascending=False)

comparison = pd.DataFrame({
    "built-in": built_in,
    "permutation": permutation
}).sort_values("built-in", ascending=False)

print("\n================================ Permutation importance with a column of random values ================================")
print(comparison)
```

**Your answers:**

a) We expect ca_1.0, ca_2.0 and thal_2.0, with odds ratios 0.139817, 0.333048 and 2.740746, respectively. From the permutation plot we can see that it's actually thal, cp and oldpeak. As the plot shows, the higher the decrease in accuracy, the higher it's permutation importance. Not only thal has the highest mean value, it also has a large possible interval. There is the possibility that ca could have a greater importance, given it's large possible interval, but it can't be proven by this plot alone.

b) KNN is most likely to show the biggest difference in prediction, as it has the highest permutation importance. The thal attribute has a permutation importance of 0.1068 for KNN, compared with 0.0816 for the decision tree, 0.0623 for random forest, and 0.0556 for logistic regression.

c) The permutation importance differs considerably between the training and test sets. In particular, ca and age have relatively high importance on the training set (0.0650 and 0.0567), but almost zero importance on the test set (-0.0014 and -0.0010). This suggests that the random forest has learned patterns involving these attributes that do not generalize well to unseen data, which is an indication of overfitting. However, thal has high importance on both sets, suggesting that its contribution is more consistent.

d) Adding a noisy copy of thalach decreases the permutation importance of the original thalach (0.062319 -> 0.024638). The two features contain very similar information, so when one is permuted, the random forest can still use the other to make similar predictions. Therefore, the prediction changes less and the measured permutation importance of thalach decreases. This demonstrates that permutation importance can be affected by correlated or redundant features.

e) The built-in feature_importances_ gives substantial importance to the random column (0.077953), while permutation importance gives it only a small importance (0.007246). This happens because the random forest's built-in importance is based on impurity reduction, so random features can sometimes appear useful due to chance splits. Permutation importance instead measures the effect of the feature on predictive performance, so a truly random feature should have an importance close to zero.

> two-way: `features=[("trestbps", "cp")], categorical_features=["cp"]`; log-odds scale: `response_method="decision_function"`

```python
features = ["thalach", "oldpeak"]  # TODO (a): choose the two attributes you think explain the most

for name, model in models.items():
    disp = PartialDependenceDisplay.from_estimator(
        model, X_test, features, kind="both", centered=True, random_state=S)
    disp.figure_.suptitle(f"PDP + ICE, {name} - group {GROUP_NUMBER}")
    plt.show()
```

```python
# TODO (b): two-way PDP of trestbps and cp
features = [("trestbps", "cp")]

for name, model in models.items():
    disp = PartialDependenceDisplay.from_estimator(
        model, X_test, features, kind="average", centered=True, random_state=S)

    disp.figure_.suptitle(f"2-way PDP, {name} - group {GROUP_NUMBER}")
    plt.show()

# TODO (c): hand-coded Ceteris Paribus profile
features = ["thalach", "cp"]

for name, model in models.items():
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))

    for ax, feature in zip(axes, features):
        values = np.sort(X_test[feature].unique())

        predictions = []

        for value in values:
            patient_modified = x_p.copy()
            patient_modified[feature] = value

            prediction = model.predict_proba(patient_modified)[0, 1]
            predictions.append(prediction)

        ax.plot(values, predictions, marker="o")
        ax.set_xlabel(feature)
        ax.set_ylabel("Prediction")
        ax.set_title(f"CP profile: {feature}")
        ax.grid()

    fig.suptitle(f"Ceteris Paribus Profiles - {name} - group {GROUP_NUMBER}")
    plt.tight_layout()
    plt.show()
```

**Your answers:**

a) We believe thalach and oldpeak are the features that better explain the models. They're both continuous, numerical data and their permutation importance if high on most models. The models behaviors vary significantly based on their underlying logic: 
- Logistic Regression imposes smooth, strictly parallel linear lines with zero interactions.
- Decision Tree produces rigid step functions at specific split threshold.
- k-Nearest Neighbors forms blocky, plateauing steps as local neighborhood sets update.
- Random Forest generates a smooth, non-linear curve by averaging an ensemble of trees.

Patients do not all behave the same way, as demonstrated by the wide fan-out of individual ICE lines in models like Random Forest, which indicates strong interaction effects with other features. Because the PDP line only displays the global average, it can disguise critical sub-group dynamics. For example, in the Decision Tree plot for thalach above 145, the PDP dashed line indicates a mild average risk increase of roughly +0.12. However, the ICE lines reveal that almost no individual patient actually has a change of +0.12: the model actually splits patients into extreme outcomes, jumping to +1.00 for one group and dropping to 0.00 or -1.00 for others, a bimodal polarization completely masked by the PDP average.   

b) Across all models, the two-way partial dependence plots demonstrate that trestbps dominates the predicted outcome, while cp has a negligible effect. The contour lines are predominantly vertical or near-vertical, meaning that varying cp along the vertical axis barely changes the predicted values. Additionally, as trestbps increases from left to right, the target prediction consistently decreases across all models.

> **Hint:** `shap.plots.beeswarm(sv)`, `shap.plots.bar(sv)`, `shap.plots.scatter(sv[:, "exang"], color=sv)`;
> `shap.plots.waterfall(sv[p])`, `shap.plots.force(sv[p])`; `sv.base_values[p]`, `sv.values[p].sum()`, `clf.decision_function(Xt_test)[p]`;

```python
shap.initjs()

# SHAP for the logistic model, on the preprocessed (scaled) data.
# The values are on the log-odds scale.
pre, clf = logreg[:-1], logreg[-1]
Xt_train, Xt_test = pre.transform(X_train), pre.transform(X_test)

# Use the full training set as background (by default SHAP keeps only 100 rows)
masker = shap.maskers.Independent(Xt_train, max_samples=len(Xt_train))
explainer = shap.LinearExplainer(clf, masker)
sv = explainer(Xt_test)

shap.plots.beeswarm(sv, max_display=20)
```
